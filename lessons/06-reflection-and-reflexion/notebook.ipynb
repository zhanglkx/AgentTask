# 第 6 章：Reflection 与 Reflexion

让 Agent 自我批评与改进——写作 agent 和 coding agent 雏形。

## 6.1 Basic Reflection

Reflection = 生成 → 批判 → 修订循环：

1. **生成**: 写初版
2. **批判**: reviewer 指出问题
3. **决策**: ACCEPT → 返回,或继续修订

类比前端：
- Generator ≈ Feature developer
- Critic ≈ Code reviewer
- ACCEPT ≈ CI passing


```
from langchain_community.chat_models.fake import FakeListChatModel

from writing_agent import run_writing_agent

gen = FakeListChatModel(responses=["v1", "v2（改进版）"])
crit = FakeListChatModel(responses=["需要改进", "ACCEPT"])


result = run_writing_agent(gen, crit, "写一段介绍", max_iterations=3)
print(f"最终版本: {result['final']}")
print(f"迭代次数: {result['iteration']}")
```

## 6.2 Reflexion

Reflexion = Reflection + 经验记忆（episodic memory）：

1. 从过去的成功/失败中**回忆**（recall）
2. 带经验**行动**（act）
3. **批判**（critic）
4. 如果失败 → **反思**（reflect）并存入经验库

类比前端：
- ExperienceStore ≈ Git log（记住过去经验）
- SUCCESS ≈ CI passing
- Reflexion ≈ Agile retrospective（从失败中学习）


```
from coding_agent import run_coding_agent

actor = FakeListChatModel(responses=["v1（有bug）", "v2（修复版）"])
critic = FakeListChatModel(responses=["FAIL: 返回 None", "SUCCESS"])


result = run_coding_agent(actor, critic, "写一个加法函数", max_iterations=3)
print(f"最终代码: {result['attempt']}")
```
